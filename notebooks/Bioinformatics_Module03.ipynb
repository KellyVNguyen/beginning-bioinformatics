````python
## Asking Gemini to give 3 examples

## Example 1: Reassigning a variable
x = 5
print(x)  # Output: 5

x = 10  # Reassigns x to 10
print(x)  # Output: 10

## Example 2: Assigning using the variable's current value
score = 100
score = score + 50  # Evaluates 100 + 50 = 150, then stores 150 in score
print(score)  # Output: 150

## Example 3: Swapping two variables
a = "apple"
b = "banana"

# Swapping values using a temporary variable
temp = a  # temp is now "apple"
a = b  # a is now "banana"
b = temp  # b is now "apple"

print(a, b)  # Output: banana apple
````

####**TASK P15.2** <font color=orange>**Complete Rosalind Problem #4 Working with Files**</font>


Given: A file containing at most 1000 lines.

Return: A file containing all the even-numbered lines from the original file. Assume 1-based numbering of lines.

````python
## Sample
## .join() list slicing approach

with open("practice.txt", "r") as data1:
    lines = data1.readlines()

# Slice every second line starting from index 1 (line 2)
even_lines = lines[1::2]

with open("output.txt", "w") as data2:
    data2.write("".join(even_lines))
````

````python
## Sample
## loop approach

with open("practice.txt", "r") as data1:
    lines = data1.readlines()

line_number = 1

with open("output.txt", "w") as data2:
    for line in lines:
        if line_number % 2 == 0:
            data2.write(line)
        line_number += 1
````
