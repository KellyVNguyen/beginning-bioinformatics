####**TASK P15.2** <font color=orange>**Complete Rosalind Problem #4 Working with Files**</font>


Given: A file containing at most 1000 lines.

Return: A file containing all the even-numbered lines from the original file. Assume 1-based numbering of lines.

````python
## Sample
## .join() list slicing approach

with open("practice.txt", "r") as data1:
    lines = data1.readlines()

# Slice every second line starting from index 1 (line 2)
even_lines = lines[1::2]

with open("output.txt", "w") as data2:
    data2.write("".join(even_lines))
````

````python
## Sample
## loop approach

with open("practice.txt", "r") as data1:
    lines = data1.readlines()

line_number = 1

with open("output.txt", "w") as data2:
    for line in lines:
        if line_number % 2 == 0:
            data2.write(line)
        line_number += 1
````

####**TASK P18.2** <font color= orange>**Complete Rosalind Problem #5  Dictionaries**</font>

Given: A string s
 of length at most 10000 letters.

Return: The number of occurrences of each word in s
, where words are separated by spaces. Words are case-sensitive, and the lines in the output can be in any order.

````python
## Sample

# steps 1 and 2
with open("practice_5.txt", "r") as data:
    text = data.read().strip()

# step 3
dictionary = {}

# steps 4 and 5
for word in text.split(" "):
    if word in dictionary:
        dictionary[word] += 1
    else:
        dictionary[word] = 1

# step 6
for key, value in dictionary.items():
    print(key, value)
````

####**TASK P19.1** <font color= orange>**Complete Rosalind Problem #6  Counting DNA Nucleotides**</font>

A string is simply an ordered collection of symbols selected from some alphabet and formed into a word; the length of a string is the number of symbols that it contains.

An example of a length 21 DNA string (whose alphabet contains the symbols 'A', 'C', 'G', and 'T') is "ATGCTTCAGAAAGGTCTTACG."

Given: A DNA string s
 of length at most 1000 nt.

Return: Four integers (separated by spaces) counting the respective number of times that the symbols 'A', 'C', 'G', and 'T' occur in s
.

````python
## sample dataset
with open("sample_6.txt", "r") as data:
  s = data.read().strip()

a_count = s.count('A')
c_count = s.count('C')
g_count = s.count('G')
t_count = s.count('T')

print(a_count, c_count, g_count, t_count)
````

####**TASK P19.2** Modify your code for the Rosalind #5 Dictionary problem so that it uses count( ) to count the words.

````python
with open("rosalind_ini6.txt", "r") as data:
    text = data.read().strip()

dictionary = {}

words = text.split(" ")

for word in words:
  if word not in dictionary:
      dictionary[word] = words.count(word)

for key, value in dictionary.items():
    print(key, value)
````
